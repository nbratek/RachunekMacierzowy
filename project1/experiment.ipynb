```python
import numpy as np
import time
import matplotlib.pyplot as plt

from project1.binet_algorithm import BinetAlgorithm
```

```python
def run_matrix_multiplication(l):
    sizes = []
    times = []
    operations = []
    for k in range(2,10):
        n = 2**k
        A = np.random.rand(n,n)
        B = np.random.rand(n,n)
        binet = BinetAlgorithm(l)
        start = time.time()
        binet.mul(A,B)
        end = time.time()
        sizes.append(n)
        total_time = end - start
        times.append(total_time)
        operations.append(binet.calc.total_count)
    return sizes, times, operations
```

```python
sizes, times, operations = run_matrix_multiplication(l=2)
plt.figure(figsize=(8,5))
plt.plot(sizes, times, marker="o")
plt.xlabel("Matrix size")
plt.ylabel("Time")
plt.title("Execution time")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(sizes, operations, marker="o")
plt.xlabel("Matrix size")
plt.ylabel("Number of operations")
plt.title("Number of operations")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8,5))
for l in [1,2,3,4]:
    sizes, times, operations = run_matrix_multiplication(l)
    plt.plot(sizes, times, marker="o", label=f"l={l}")

plt.xlabel("Matrix size")
plt.ylabel("Time")
plt.title("Time for different l")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8,5))
for l in [1,2,3,4]:
    sizes, times, operations = run_matrix_multiplication(l)
    plt.plot(sizes, operations, marker="o", label=f"l={l}")

plt.xlabel("Matrix size")
plt.ylabel("Operations")
plt.title("Operations for different l")
plt.legend()
plt.grid()
plt.show()
```
